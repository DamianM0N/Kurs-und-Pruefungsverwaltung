### Regeln:
1. Punktzahl Projekt, Punktzahl Klausur $\in [0,50]$ und Punktzahlen können nur natürliche Zahlen sein.
2. Gesamtpunktzahl = Punktzahl Projekt + Punktzahl Klausur (maximal 100 Punkte).
3. Prüfung ist nur bestanden, falls Gesamtpunktzahl mindestens 50 Punkte beträgt und Klausur Punkte mindestens 25 Punkte sind.
4. Fehlende Prüfungsbestandteile werden als fehlender Dictionary-Schlüssel dargestellt. Solche Datensätze gelten als unvollständig und werden aus Durchschnitt, Bestehensquote und Notenverteilung ausgeschlossen.
5. Gesamtpunkte, Bestehensstatus und Note werden nicht dauerhaft im Dictionary gespeichert, sondern bei Bedarf aus den aktuellen Teilpunkten berechnet. Nach einer Nachkorrektur entstehen dadurch automatisch die aktuellen Ergebnisse.

```python
def id_vorhanden(id):
    '''
    Funktion die untersucht, ob eine Studenten-ID bereits in der Liste vorhanden ist
    '''
    
    #Überprüfung ob es sich um eine natürliche Zahl größer 101 handelt
    if (id.isdigit() == False) or (int(id) < 101):
        return 'ungültige Zahl'
    #Überprüfung ob ID kleiner ist als kleinste erlaubte ID
    else:
        for student in studierende:
            if student['id'] == int(id):
                return True
        return False
```

```python
def pruefe_punktzahl(punktzahl):
    '''
    Funktion die überprüft ob die eingetragene Punktzahl die angegebenen Regeln erfüllt
    '''

    if (punktzahl.isdigit() == False) or not (0 <= int(punktzahl) <= 50):
        return False
    else:
        return True
```

```python
def pruefe_name(name):
    '''
    Funktion die überprüft ob eingetragener Name zulässig ist. Ein Name ist zulässig, falls dieser nur aus Buchstaben besteht. Bindestriche für Doppelnamen ist erlaubt.
    '''

    #Überprüfung ob der Name leer ist
    if name == '':
        return False
    #Überprüfung ob es ein Leerzeichen gibt oder der Name nicht aus Buchstaben besteht
    for zeichen in name:
        if (zeichen == ' ') or (zeichen.isalpha() == False and zeichen != '-'):
            return False
    return True
```

```python
def suche_studenten(id):
    '''
    Funktion die überprüft ob es einen Studenten zur gegebenen Studenten-ID gibt.
    '''
    
    ergebnis = id_vorhanden(id)
    if ergebnis == 'ungültige Zahl':
        print('Fehler: Studenten-ID muss eine natürliche Zahl größer als 100 sein.')
        return None
    elif ergebnis == False:
        print('Kein Student mit dieser ID wurde gefunden.')
        return None
    else:
        for student in studierende:
            if student['id'] == int(id):
                return student
```

```python
def student_hinzufuegen(id, name, projekt_punkte, klausur_punkte):
    '''
    Funktion die einen neuen Studenten hinzufügt, falls ID nicht vergeben wurde und Punkte die Regeln erfüllen.
    '''

    #Überprüfung ob Eingabe der ID korrekt ist und ob ID noch frei ist
    ergebnis = id_vorhanden(id)
    if ergebnis == 'ungültige Zahl':
        print('Fehler: Studenten-ID muss eine natürliche Zahl größer als 100 sein.')
        return None
    elif ergebnis == True:
        print('Fehler: Studenten-ID ist bereits vergeben.')
        return None

    #Überprüfung ob eingegebener Name erlaubt ist
    if pruefe_name(name) == False:
        print('Fehler: Der Name darf nicht leer sein und darf nur aus Buchstaben, sowie Bindestrichen bestehen.')
        return None

    #Überprüfung ob Punktzahlen korrekt sind
    if not (pruefe_punktzahl(projekt_punkte) == True) or not (pruefe_punktzahl(klausur_punkte) == True):
        print('Punkte müssen zwischen 0 und 50 liegen und dürfen keine Kommazahlen sein.')
        return None

    #Falls alle Eingaben korrekt sind, wird ein neuer Student hinzugefügt
    studierende.append({'id': int(id), 'name': name, 'projekt': int(projekt_punkte), 'klausur': int(klausur_punkte)})
```

```python
def bearbeite_name():
    '''
    Funktion die einen Namen zur gegebenen ID überarbeitet
    '''
    
    id = input('Gib die ID des Studenten ein, die du bearbeiten möchtest: ')
    student = suche_studenten(id)
    if student is not None:
        print(f'Die Daten zu dieser ID lauten: {student}\n')
    else:
        return
    name = input('Ändere Namen zu: ')
    if pruefe_name(name) == False:
        print('Fehler: Der Name darf nicht leer sein und darf nur aus Buchstaben, sowie Bindestrichen bestehen.')
    else:
        student['name'] = name
    
```

```python
def bearbeite_punkte():
    '''
    Funktion die Punkte von dem Projekt oder von der Prüfung überarbeiten soll
    '''

    #Suche nach der ID
    id = input('Gib die ID des Studenten ein, die du bearbeiten möchtest: ')
    student = suche_studenten(id)
    if student is not None:
        print(f'Die Daten zu dieser ID lauten: {student}\n')
    else:
        return

    #Ändere Punktzahl falls Punktzahl valide
    punkte = input('Gib deine neue Punktzahl ein: ')
    if pruefe_punktzahl(punkte) == True:
        typ = input('Soll die Punktzahl für [Projekt] oder für [Klausur] geändert werden: ')
        if typ.lower() == 'projekt':
            student['projekt'] = punkte
        elif typ.lower() == 'klausur':
            student['klausur'] = punkte
        else:
            print('Fehler: Überprüfe Eingabe. Es wird nur Projekt und Klausur akzeptiert.')
    else:
        print('Punkte müssen zwischen 0 und 50 liegen und dürfen keine Kommazahlen sein.')
        return None
```

```python
def bearbeite_id():
    '''
    Funktion die die ID eines Studenten ändert
    '''

    #Suche nach der ID
    id = input('Gib die ID des Studenten ein, die du bearbeiten möchtest: ')
    student = suche_studenten(id)
    if student is not None:
        print(f'Die Daten zu dieser ID lauten: {student}\n')
    else:
        return

    #Wahl einer neuen ID
    neue_id = input('Gib die neue ID ein: ')
    if (neue_id.isdigit() == False) or (int(neue_id) < 101):
        print('Fehler: Studenten-ID muss eine natürliche Zahl größer 100 sein.')
    else:
        if id_vorhanden(neue_id) == True:
            print('Fehler: Die Studenten-ID ist bereits vergeben.')
        else:
            student['id'] = neue_id
```

```python
def berechne_note(id = None, fehlermeldung = True):
    '''
    Funktion welche die gesamte Punktzahl und die Note eines Studenten berechnet
    '''

    #Studenten-ID wird gesucht
    if id == None:
        id = input('Gib die ID des Studenten ein, für welchen du die Punkte und Note berechnen möchtest: ')
    student = suche_studenten(id)
    if student is not None:
        endnote = 0.0
        gesamt_punkte = 0.0
        #Überprüfung ob in sowohl bei Klausur als auch Projekt Punkte stehen
        try:
            projekt_punkte = student['projekt']
            klausur_punkte = student['klausur']
            gesamt_punkte = projekt_punkte + klausur_punkte
            #Bedingung fürs Bestehen wird überprüft
            if (klausur_punkte < 25) or (gesamt_punkte < 50):
                endnote = 5.0
                return endnote, gesamt_punkte 
            else:
                for noten in notentabelle:
                    if gesamt_punkte in noten['gesamtpunkte']:
                        endnote = noten['note']
                        return endnote, gesamt_punkte
        except KeyError:
            if fehlermeldung == True:
                print('Fehler: Dieser Student hat nicht in beiden Bereichen Punkte eingetragen.')
```

```python
def berechne_kursdurchschnitt():
    '''
    Funktion welche den Kursdurchschnitt berechnet
    '''
    kurs_note = 0.0
    anzahl_studierender = 0

    #Berechnung der Noten für jeden Student aus der Liste, der in beiden Einträgen Punkte hat
    for student in studierende:
        ergebnis = berechne_note(str(student['id']), False)
        if ergebnis is not None:
            endnote, gesamtpunkte = ergebnis
            kurs_note += endnote
            anzahl_studierender += 1
    if anzahl_studierender > 0:
        kurs_note = kurs_note / anzahl_studierender
        print(f'Die durchschnittliche Note des Kurses beträgt {kurs_note:.2f}')
    else:
        print('Fehler: Es konnte keine Durchschnittsnote berechnet werden.')
```

```python
def berechne_notenverteilung():
    '''
    Funktion die die Notenverteilung berechnet, speichert und als Text ausgibt
    '''

    for note in notentabelle:
        note['anzahl_studenten'] = 0

    for student in studierende:
        ergebnis = berechne_note(str(student['id']), False)

        for note in notentabelle:
            if ergebnis != None:
                if ergebnis[1] in note['gesamtpunkte']:
                    note['anzahl_studenten'] += 1
            else:
                continue

    for note in notentabelle:
        print(f'• {note['anzahl_studenten']} haben eine {note['note']}')
```

```python
def berechne_bestehensquote():
    '''
    Funktion die die Bestehensquote berechnet
    '''

    anzahl_bestanden = 0
    anzahl_nicht_bestanden = 0

    for student in studierende:
        ergebnis = berechne_note(str(student['id']), False)
        if ergebnis != None:
            if ergebnis[0] != 5.0:
                anzahl_bestanden += 1
            else:
                anzahl_nicht_bestanden += 1
        else: 
            continue

    if anzahl_bestanden + anzahl_nicht_bestanden != 0:
        print(f'Die Bestehensquote liegt bei {((anzahl_bestanden)/(anzahl_bestanden + anzahl_nicht_bestanden)) * 100:.2f}%')
    else:
        print(f'Fehler: Bestehensquote konnte nicht ermittelt werden.')
```

```python
def finde_besten_studenten():
    '''
    Funktion, die den oder die besten Studenten anhand der Punkt findet
    '''

    max_punkte = -1
    beste_studenten = []

    for student in studierende:

        try:
            projekt_punkte = student['projekt']
            klausur_punkte = student['klausur']
        except KeyError:
            continue

        gesamtpunkte = klausur_punkte + projekt_punkte

        if gesamtpunkte > max_punkte:
            max_punkte = gesamtpunkte
            beste_studenten = [student]
        elif gesamtpunkte == max_punkte:
            beste_studenten.append(student)

    if len(beste_studenten) > 0:
        print(f'Die höchste Punktzahl beträgt {max_punkte} und wurde erreicht von:')

        for student in beste_studenten:
            print(f'• {student['name']} | ID: {student['id']}')
    else:
        print('Fehler: Es gibt keinen Studenten mit vollständigen Punktzahlen.')
```
